```python
import cv2
import numpy as np

image = cv2.imread("image.png")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_rgb_new = image_rgb.copy()

image_rgb_new[:, :, 0] = (
    (image_rgb_new[:, :, 0] - np.min(image_rgb_new[:, :, 0]))
    / (np.max(image_rgb_new[:, :, 0]) - np.min(image_rgb_new[:, :, 0]))
    * 255
)
image_rgb_new[:, :, 1] = (
    (image_rgb_new[:, :, 1] - np.min(image_rgb_new[:, :, 1]))
    / (np.max(image_rgb_new[:, :, 1]) - np.min(image_rgb_new[:, :, 1]))
    * 255
)
image_rgb_new[:, :, 2] = (
    (image_rgb_new[:, :, 2] - np.min(image_rgb_new[:, :, 2]))
    / (np.max(image_rgb_new[:, :, 2]) - np.min(image_rgb_new[:, :, 2]))
    * 255
)
print("R,G,B:", image_rgb_new[422][644])
```
