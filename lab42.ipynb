```python
import cv2
from matplotlib import pyplot as plt
import numpy as np

img = cv2.imread("image.png")
image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
laplacian = cv2.Laplacian(image_rgb, -1)
plt.imshow(laplacian)
sum_intensity_laplacian = np.sum(laplacian)
print(
    f"Сумма интенсивностей всех пикселей по всем трем каналам: {sum_intensity_laplacian}"
)
canny = cv2.Canny(image_rgb, 100, 200)
plt.imshow(canny, cmap="gray")
sum_intensity_canny = np.sum(canny)
print(f"Сумма интенсивностей всех пикселей: {sum_intensity_canny}")
```
