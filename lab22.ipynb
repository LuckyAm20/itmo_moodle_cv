```python
import cv2
import matplotlib.pyplot as plt
```

```python
img_bgr = cv2.imread("image.png")
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
crop = img_rgb[1254:1764, 2392:3042]

plt.figure(figsize=(8, 8))
plt.imshow(crop)
plt.axis("off")
plt.show()
```
