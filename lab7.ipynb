```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

%matplotlib inline
import matplotlib.image as mpimg
import matplotlib as mpl
from PIL import Image

mpl.rcParams["figure.dpi"] = 200
```

```python
image = cv2.imread("image.png")

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
height, width, _ = image_rgb.shape
print("Высота изображения в пикселях", height)
print("Ширина изображения в пикселях", width)
print("Среднее значение интенсивности пикселя", round(image_rgb.mean(), 3))
```

```python
def exposure(image, p):
    return image.copy() * (2**p)


print(
    "Cреднее значение интенсивности пикселя преобразованного изображения при p = -0.47",
    round(np.mean(exposure(image_rgb, -0.47)), 3),
)
def exposure(image, p):
    img = image.astype(np.float64) * (2**p)
    img = np.clip(img, 0, 255)
    img = np.rint(img).astype(np.uint8)
    return img
print(
    "Cреднее значение интенсивности пикселя преобразованного изображения при p = 0.5",
    np.mean(exposure(image_rgb, 0.5)),
)
```

```python
def contrast(image, p):
    img_result = image.copy() / 255.0
    if p > 0:
        img_result = (img_result - 0.5) / (1.0 - p)
    else:
        img_result = (img_result - 0.5) * (1.0 - p)
    for i in range(img_result.shape[0]):
        for j in range(img_result.shape[1]):
            for k in range(img_result.shape[2]):
                if img_result[i][j][k] > 1:
                    img_result[i][j][k] = 1
                elif img_result[i][j][k] < 0:
                    img_result[i][j][k] = 0
                img_result[i][j][k] = round(img_result[i][j][k] * 255.0, 0)
    img_result = img_result.astype(int)
    return img_result


img_1 = image_rgb.copy()
img_2 = image_rgb.copy()
print(
    "Cреднее значение интенсивности пикселя преобразованного изображения при p = -0.47",
    round(np.mean(contrast(img_1, -0.47)), 3),
)
print(
    "Cреднее значение интенсивности пикселя преобразованного изображения при p = 0.5",
    round(np.mean(contrast(img_2, 0.5)), 3),
)
```
