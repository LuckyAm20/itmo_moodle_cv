```python
import cv2
import os

img_bgr = cv2.imread("image.png")
height, width = img_bgr.shape[:2]

ratio = width / height
new_width = 620
new_height = int(new_width / ratio)

img_resized_bgr = cv2.resize(
    img_bgr,
    (new_width, new_height),
    interpolation=cv2.INTER_AREA
)

img_resized_rgb = cv2.cvtColor(img_resized_bgr, cv2.COLOR_BGR2RGB)

print("ratio =", round(ratio, 3))
print("new_height =", new_height)
print("nbytes =", img_resized_rgb.nbytes)

cv2.imwrite("out.png", img_resized_rgb)
print("png size =", os.path.getsize("out.png"))

cv2.imwrite("out.jpg", img_resized_rgb)
print("jpg size =", os.path.getsize("out.jpg"))

cv2.imwrite("out.bmp", img_resized_rgb)
print("bmp size =", os.path.getsize("out.bmp"))
```

```python
import cv2
import os

img_bgr = cv2.imread("image.png")
height, width = img_bgr.shape[:2]

ratio = width / height

new_width = 520
new_height = int(new_width / ratio)

img_resized_bgr = cv2.resize(
    img_bgr,
    (new_width, new_height),
    interpolation=cv2.INTER_AREA
)

img_resized_rgb = cv2.cvtColor(img_resized_bgr, cv2.COLOR_BGR2RGB)

output_path = "resized_rgb.jpg"
cv2.imwrite(output_path, cv2.cvtColor(img_resized_rgb, cv2.COLOR_RGB2BGR))

print(round(ratio, 3))
print(new_height)
print(os.path.getsize(output_path))
```
