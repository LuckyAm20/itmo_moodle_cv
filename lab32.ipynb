```python
import cv2
import numpy as np

image = cv2.imread("image.png")
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

max_density_red = np.max(
    np.histogram(image_rgb[:, :, 0].ravel(), bins=256, density=True)[0]
)
max_density_green = np.max(
    np.histogram(image_rgb[:, :, 1].ravel(), bins=256, density=True)[0]
)
max_density_blue = np.max(
    np.histogram(image_rgb[:, :, 2].ravel(), bins=256, density=True)[0]
)

max_density = max(max_density_red, max_density_green, max_density_blue)
print("Наибольшее значение плотности среди трех каналов:", round(max_density, 4))
```
